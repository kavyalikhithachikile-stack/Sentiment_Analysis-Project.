```python
# SENTIMENT ANALYSIS PROJECT
# Classifies text as Positive, Negative, or Neutral

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

# Step 1: Create a sample dataset
data = {
    "text": [
        "I love this product",
        "The food was delicious",
        "Amazing service and friendly staff",
        "This movie was wonderful",
        "I am very happy with my purchase",
        "Excellent quality and great value",
        "The experience was fantastic",
        "I really enjoyed this restaurant",
        "Absolutely brilliant product",
        "The delivery was fast and perfect",
        "I hate this product",
        "The food was terrible",
        "Very poor customer service",
        "This movie was boring",
        "I am disappointed with my purchase",
        "The quality is awful",
        "The experience was horrible",
        "I regret buying this product",
        "The restaurant was dirty",
        "The delivery was late and disappointing",
        "The product arrived today",
        "The restaurant is near my house",
        "The package contains two items",
        "The shop opens at nine",
        "I purchased this yesterday",
        "The movie is two hours long",
        "The order number is 12345",
        "The phone has a six inch screen",
        "The meeting starts at ten",
        "The product comes in three colors"
    ],
    "sentiment": [
        "Positive", "Positive", "Positive", "Positive", "Positive",
        "Positive", "Positive", "Positive", "Positive", "Positive",
        "Negative", "Negative", "Negative", "Negative", "Negative",
        "Negative", "Negative", "Negative", "Negative", "Negative",
        "Neutral", "Neutral", "Neutral", "Neutral", "Neutral",
        "Neutral", "Neutral", "Neutral", "Neutral", "Neutral"
    ]
}

df = pd.DataFrame(data)

# Step 2: Split the dataset into training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    df["text"],
    df["sentiment"],
    test_size=0.2,
    random_state=42,
    stratify=df["sentiment"]
)

# Step 3: Build the machine-learning pipeline
model = Pipeline([
    ("tfidf", TfidfVectorizer(
        lowercase=True,
        ngram_range=(1, 2),
        stop_words="english"
    )),
    ("classifier", LogisticRegression(max_iter=1000))
])

# Step 4: Train the model
model.fit(X_train, y_train)

# Step 5: Evaluate the model
predictions = model.predict(X_test)

print("SENTIMENT ANALYSIS RESULTS")
print("-" * 35)
print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
print("Accuracy:", round(accuracy_score(y_test, predictions) * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(y_test, predictions, zero_division=0))

# Step 6: Test your own sentences
print("\nTry your own sentences!")
print("Type 'exit' to stop.")

while True:
    sentence = input("\nEnter a sentence: ")

    if sentence.strip().lower() == "exit":
        print("Thank you for using Sentiment Analysis!")
        break

    if not sentence.strip():
        print("Please enter some text.")
        continue

    result = model.predict([sentence])[0]
    print("Predicted Sentiment:", result)
```
